```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully")
```

```python
df = pd.read_csv("../Data/raw/healthcare_dataset.csv")
```

```python
clean_df = df.copy()
```

```python
clean_df.shape
```

```python
clean_df.isnull().sum()
```

```python
missing = clean_df.isnull().sum()

missing_percent = (
    missing / len(clean_df)
) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percent
})

missing_df.sort_values(
    "Percentage",
    ascending=False
)
```

```python
clean_df = clean_df.drop_duplicates()
```

```python
clean_df.dtypes
```

```python
clean_df["Date of Admission"] = pd.to_datetime(
    clean_df["Date of Admission"]
)

clean_df["Discharge Date"] = pd.to_datetime(
    clean_df["Discharge Date"]
)
```

```python
clean_df.info()
```

```python
clean_df.describe()
```

```python
clean_df.describe(include="object")
```

```python
clean_df.to_csv(
    "../Data/cleaned/healthcare_cleaned.csv",
    index=False
)
```
