## Objective

Create new business-friendly columns for better analysis in SQL and Power BI.

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully")
```

```python
df = pd.read_csv("../Data/cleaned/healthcare_cleaned.csv")
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df["Length of Stay"] = (
    df["Discharge Date"] -
    df["Date of Admission"]
).dt.days
```

```python
df[["Date of Admission","Discharge Date","Length of Stay"]].head()
```

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0,18,35,50,65,100],
    labels=[
        "Child",
        "Young Adult",
        "Adult",
        "Senior Adult",
        "Elderly"
    ]
)
```

```python
df["Age Group"].value_counts()
```

```python
df["Admission Year"] = df["Date of Admission"].dt.year
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month_name()
```

```python
df["Admission Day"] = df["Date of Admission"].dt.day_name()
```

```python
df["Billing Category"] = pd.qcut(
    df["Billing Amount"],
    q=3,
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)
```

```python
df["Billing Category"].value_counts()
```

```python
df.to_csv(
    "../Data/cleaned/healthcare_feature_engineered.csv",
    index=False
)
```
